```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv("Nutrition__obesity.csv")
df.head()
```

```python
df.info()
df.describe()
```

```python
X = df[['Sample_Size']]
y = df['Data_Value']
```

```python
plt.scatter(X, y)
plt.xlabel("Sample Size")
plt.ylabel("Obesity (%)")
plt.title("Scatter Plot")
plt.show()
```

```python
data = df[['Sample_Size', 'Data_Value']].dropna()

X = data[['Sample_Size']]
y = data['Data_Value']
```

```python
model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
y_pred = model.predict(X)

plt.scatter(X, y)
plt.plot(X, y_pred)

plt.xlabel("Sample Size")
plt.ylabel("Obesity (%)")
plt.title("Linear Regression Result")

plt.show()
```

```python
import numpy as np

new_sample = np.array([[2500]])
prediction = model.predict(new_sample)

print("Prediksi Data_Value:", prediction[0])
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(y, y_pred))
print("RMSE:", rmse)
```
